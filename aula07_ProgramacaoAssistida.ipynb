Eh primo

```python
import math

def eh_primo(n):
  # Números menores que 2 não são primos.
  if n < 2:
    return False
  # 2 é o único número primo par.
  if n == 2:
    return True
  # Números pares maiores que 2 não são primos.
  if n % 2 == 0:
    return False
  # Verifica divisores ímpares de 3 até a raiz quadrada de n para otimizar.
  # Se n tem um divisor d, então n = d * k. Se d > sqrt(n), então k < sqrt(n).
  # Assim, só precisamos verificar até sqrt(n).
  for i in range(3, int(math.sqrt(n)) + 1, 2):
    if n % i == 0:
      return False
  return True
```

inverção string

```python
def inverte_texto(texto): # Define uma função chamada 'inverte_texto' que recebe um argumento 'texto'.
  invertido ='' # Inicializa uma string vazia chamada 'invertido'. Esta string será usada para construir o texto invertido.
  for letra in texto: # Inicia um loop que percorre cada caractere (chamado 'letra') na string 'texto' de entrada.
    invertido = letra + invertido # Adiciona o caractere atual ('letra') no início da string 'invertido', invertendo assim a ordem.
  return invertido # Retorna a string 'invertido' após o loop ter processado todos os caracteres.
```

Calculadora Simples

```python
def calculadora(a, b, operacao):
    if operacao == "+":
        return a + b
    elif operacao == "-":
        return a - b
    elif operacao == "*":
        return a * b
    elif operacao == "/":
        if b != 0:
            return a / b
        else:
            return "Erro: divisão por zero"
    else:
        return "Operação inválida"


# Exemplo de uso
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
op = input("Digite a operação (+, -, *, /): ")

resultado = calculadora(num1, num2, op)
print("Resultado:", resultado)
```
