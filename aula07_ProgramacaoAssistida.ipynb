```python
numeros = [1, 3, 6 ,7 ,8] # Define uma lista de números.
soma = 0 # Inicializa a variável 'soma' com 0. Esta variável irá armazenar a soma dos números.
for num in numeros: # Inicia um loop que itera sobre cada 'num' (número) na lista 'numeros'.
  soma += num # Adiciona o 'num' atual à variável 'soma'.
print(soma) # Imprime o valor final da 'soma' após o loop ter percorrido todos os números.
```
